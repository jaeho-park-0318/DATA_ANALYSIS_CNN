```python
import csv
import matplotlib.pyplot as plt
flow = []
with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    for row in list(csv.reader(f))[1:]:
        if len(row) > 11 and row[11].strip():
            try: flow.append(float(row[11]))
            except ValueError: pass
plt.plot(flow)
plt.title("Gas flow")
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

values = []

with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    header = next(data)

    print("CSV 열 목록:", header)

    for row in data:
        if len(row) > 11:
            if "depo" in row[4].strip().lower():
                try:
                    values.append(float(row[11]))
                except ValueError:
                    continue

print("추출된 데이터 개수:", len(values))

if values:
    plt.figure(figsize=(10, 5))
    plt.plot(values, 'b.-')
    plt.title("Depo Gas Flow")
    plt.xlabel("Sample")
    plt.ylabel("Gas Flow")
    plt.grid(True)
    plt.show()
else:
    print("데이터가 없습니다. row[4]와 row[11]을 확인하세요.")
```
