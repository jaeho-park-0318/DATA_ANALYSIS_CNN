```python
import csv
f=open('seoul.csv')
data=csv.reader(f)
header=next(data)

max_temp=-999
max_date=''

for row in data :
    if row[-1]=='' :
        row[-1]=-999
    row[-1]=float(row[-1])

    if max_temp<row[-1] :
        max_temp=row[-1]
        max_date=row[0]
    
print('최고 기온일:', max_date, '최고 온도:', max_temp)
f.close()
```

```python
import csv
f=open('data_sample.csv')
data=csv.reader(f)
header=next(data)

max_heater_temp=0
min_heater_temp=1000

for row in data :
    if row[4]!='Idle' :
        if max_heater_temp<float(row[29]) :
            max_heater_temp=float(row[29])
        if min_heater_temp>float(row[29]) :
            min_heater_temp=float(row[29])
print('최소히터온도: ' , min_heater_temp, '최대히터온도: ' ,max_heater_temp)
f.close()
```
