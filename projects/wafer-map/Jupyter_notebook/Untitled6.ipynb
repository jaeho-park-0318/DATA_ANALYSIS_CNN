```python
import random
import matplotlib.pyplot as plt

dice = []

# 주사위 10,000번 던지기
for i in range(10000):
    dice.append(random.randint(1, 6))

# 히스토그램
plt.figure(figsize=(10, 5))
plt.hist(dice, bins=6, range=(0.5, 6.5),
         color='skyblue', edgecolor='black')

plt.title('Dice Simulation (10000 Rolls)')
plt.xlabel('Dice Number')
plt.ylabel('Frequency')
plt.xticks(range(1, 7))
plt.grid(axis='y')

plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

# Wafer 1~25번 데이터 저장
wf = [[] for i in range(25)]

with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    next(data)

    for row in data:
        try:
            wafer = int(float(row[7]))
            voltage = float(row[27])

            if 1 <= wafer <= 25:
                wf[wafer - 1].append(voltage)

        except (ValueError, IndexError):
            continue

# 5행 5열 그래프 생성
fig, axs = plt.subplots(5, 5, figsize=(18, 15))

for i in range(25):
    row = i // 5
    col = i % 5

    if wf[i]:
        axs[row, col].hist(
            wf[i],
            bins=50,
            edgecolor='black'
        )

    axs[row, col].set_title(f"Wafer {i+1}")
    axs[row, col].set_xlabel("Voltage")
    axs[row, col].set_ylabel("Frequency")
    axs[row, col].grid(axis='y')

plt.suptitle("Heater Voltage EDGE by Wafer", fontsize=18)
plt.tight_layout()
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

jan = []
aug = []

with open("seoul.csv", encoding="cp949", newline="") as f:
    data = csv.reader(f)
    next(data)

    for row in data:
        if len(row) < 5:
            continue

        try:
            date = row[0].strip().split("-")
            month = int(date[1])
            high = float(row[-1])

            if month == 1:
                jan.append(high)

            elif month == 8:
                aug.append(high)

        except (ValueError, IndexError):
            continue

print("1월 데이터 개수:", len(jan))
print("8월 데이터 개수:", len(aug))

plt.figure(figsize=(12, 6))

plt.hist(jan, bins=100, color='blue',
         alpha=0.6, label='January')

plt.hist(aug, bins=100, color='red',
         alpha=0.6, label='August')

plt.title('January vs August Maximum Temperature')
plt.xlabel('Maximum Temperature (°C)')
plt.ylabel('Frequency')

plt.legend()
plt.grid(axis='y')
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

TEMP_300 = []
TEMP_400 = []
TEMP_500 = []

with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    header = next(data)

    for row in data:
        try:
            if row[2] != '':
                temp = row[2].split('_')[1]

                if temp == 'TEMP300':
                    TEMP_300.append(float(row[29]))

                elif temp == 'TEMP400':
                    TEMP_400.append(float(row[29]))

                elif temp == 'TEMP500':
                    TEMP_500.append(float(row[29]))

        except (ValueError, IndexError):
            continue

print("TEMP300 데이터 개수:", len(TEMP_300))
print("TEMP400 데이터 개수:", len(TEMP_400))
print("TEMP500 데이터 개수:", len(TEMP_500))

plt.figure(figsize=(12, 6))

plt.hist(TEMP_300, label='TEMP300', color='r')
plt.hist(TEMP_400, label='TEMP400', color='g')
plt.hist(TEMP_500, label='TEMP500', color='b')

plt.title('Heater_TEMP')
plt.xlabel('Heater Temperature')
plt.ylabel('Frequency')

plt.legend(loc=2)
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

# 1월~12월 데이터를 저장할 리스트 12개 생성
month = [[] for i in range(12)]

with open("seoul.csv", encoding="cp949", newline="") as f:
    data = csv.reader(f)
    header = next(data)

    for row in data:
        try:
            # 날짜에서 월 추출
            date = row[0].strip().split("-")
            m = int(date[1])

            # 최고기온 추출
            high = float(row[-1])

            # 해당 월 리스트에 저장
            if 1 <= m <= 12:
                month[m - 1].append(high)

        except (ValueError, IndexError):
            continue

# 월별 데이터 개수 확인
for i in range(12):
    print(f"{i+1}월: {len(month[i])}개")

# Boxplot 그리기
plt.figure(figsize=(12, 6))

plt.boxplot(month)

plt.title("Monthly Maximum Temperature in Seoul")
plt.xlabel("Month")
plt.ylabel("Maximum Temperature (°C)")

plt.xticks(range(1, 13))
plt.grid(axis='y')

plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

heater_temp_set = []
heater_temp = []
diff = []

with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    next(data)

    for row in data:
        try:
            day = int(row[1].split(' ')[0].split('-')[-1])

            if day == 20:
                set_temp = float(row[28])
                measure_temp = float(row[29])

                heater_temp_set.append(set_temp)
                heater_temp.append(measure_temp)
                diff.append(set_temp - measure_temp)

        except (ValueError, IndexError):
            continue

print("20일 데이터 개수:", len(diff))

plt.figure(figsize=(14, 6))

# 동일한 Y축 사용
plt.plot(heater_temp_set, 'k--', label='Set')
plt.plot(heater_temp, 'b-', label='Measure')
plt.plot(diff, 'r-', label='diff', linewidth=2)

plt.title('Heater Temperature - Day 20')
plt.xlabel('Measurement Order')
plt.ylabel('Temperature (°C)')

# diff가 음수여도 표시되도록 범위 설정
plt.ylim(min(-20, min(diff, default=0) - 5),
         max(550, max(heater_temp_set, default=0) + 10))

plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. CSV 파일 불러오기
df = pd.read_csv('age.csv', encoding='cp949', index_col=0)

# 2. 연령별 인구수를 전체 인구수로 나누어 비율로 변환
df = df.div(df['2024년05월_계_총인구수'], axis=0)

# 3. 불필요한 열 삭제
df = df.drop(columns=[
    '2024년05월_계_총인구수',
    '2024년05월_계_연령구간인구수'
])

# 4. 결측값 제거
df1 = df.dropna(axis=0)

# 5. 비교할 지역 입력
name = input('원하는 지역의 이름을 입력해 주세요: ')

a = df1.index.str.contains(name, regex=False)

if not a.any():
    raise ValueError('해당 지역을 찾을 수 없습니다.')

df2 = df1.loc[a].iloc[0]

print('선택한 지역:', df2.name)

# 6. 지역별 인구 구조 차이 계산
x = df1.sub(df2)
y = np.power(x, 2)
z = y.sum(axis=1)

# 7. 가장 비슷한 지역 5개 추출
i = z.sort_values().index[:5]

print('\n인구 구조가 가장 비슷한 5개 지역:')
print(z.loc[i])

# 8. 그래프 그리기
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

df1.loc[i].T.plot(figsize=(14, 6))

plt.title('인구 구조가 가장 비슷한 5개 지역')
plt.xlabel('연령')
plt.ylabel('인구 비율')
plt.legend()
plt.grid(True)
plt.show()
```
