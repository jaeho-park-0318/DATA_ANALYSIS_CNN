```python
import csv
max_temp = float("-inf")
max_date = ""
with open("seoul.csv", encoding="cp949", newline="") as f:
    data = csv.reader(f)
    next(data)
    for row in data:
        if not row or not row[-1].strip(): continue
        try: temp = float(row[-1])
        except ValueError: continue
        if temp > max_temp:
            max_temp, max_date = temp, row[0]
print("최고 기온일:", max_date, "최고 기온:", max_temp)
```

```python
import csv
import matplotlib.pyplot as plt
set_values, flow_values = [], []
with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    next(data)
    for row in data:
        if len(row) <= 11: continue
        try:
            wafer = float(row[0])
            set_value = float(row[10])
            flow_value = float(row[11])
        except ValueError: continue
        if wafer > 0:
            set_values.append(set_value)
            flow_values.append(flow_value)
plt.plot(set_values, label="N2O Set")
plt.plot(flow_values, label="N2O Flow")
if set_values:
    plt.plot([v * 1.1 for v in set_values], "r--", label="Upper (example +10%)")
    plt.plot([v * 0.9 for v in set_values], "b--", label="Lower (example -10%)")
plt.legend()
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

years = []
highs = []

with open("seoul.csv", encoding="cp949", newline="") as f:
    data = csv.reader(f)
    next(data)

    for row in data:
        if len(row) < 5:
            continue

        date = row[0].strip().split("-")

        if len(date) == 3:
            if int(date[1]) == 7 and int(date[2]) == 5:
                try:
                    high = float(row[-1])

                    years.append(int(date[0]))
                    highs.append(high)

                except ValueError:
                    continue

print("추출된 데이터 개수:", len(highs))

plt.figure(figsize=(12, 5))
plt.plot(years, highs, 'r.-')

plt.title("July 5 Maximum Temperature")
plt.xlabel("Year")
plt.ylabel("Maximum Temperature (°C)")
plt.grid(True)

plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

years = []
highs = []
lows = []

with open("seoul.csv", encoding="cp949", newline="") as f:
    data = csv.reader(f)
    next(data)

    for row in data:
        if len(row) < 5:
            continue

        date = row[0].strip().split("-")

        if len(date) == 3:
            if int(date[1]) == 7 and int(date[2]) == 5:
                try:
                    high = float(row[-1])
                    low = float(row[-2])

                    years.append(int(date[0]))
                    highs.append(high)
                    lows.append(low)

                except ValueError:
                    continue

print("추출된 데이터 개수:", len(years))

plt.figure(figsize=(12, 5))

plt.plot(years, highs, 'r-', label='High Temperature')
plt.plot(years, lows, 'b-', label='Low Temperature')

plt.title("July 5 Temperature Changes")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
import csv
import matplotlib.pyplot as plt

x = []
y = []
UL = []
LL = []

with open("data_sample.csv", encoding="utf-8-sig", newline="") as f:
    data = csv.reader(f)
    header = next(data)

    for row in data:
        try:
            wafer = int(row[7])
            n2o_set = int(float(row[12]))
            n2o_flow = float(row[13])

            if wafer > 0 and n2o_set == 4000 and n2o_flow > 3000:
                x.append(wafer)
                y.append(n2o_flow)
                UL.append(n2o_set + 15)
                LL.append(n2o_set - 15)

        except (ValueError, IndexError):
            continue

plt.figure(figsize=(12, 6))

plt.plot(x, y, 'g-o', label='ACTUAL')
plt.plot(x, UL, 'r--', label='UPPER LIMIT', linewidth=3)
plt.plot(x, LL, 'r--', label='LOWER LIMIT', linewidth=3)

plt.title('N2O=4000')
plt.xlabel('Wafer Number')
plt.ylabel('N2O Flow')

plt.xticks(range(1, 26))
plt.legend(loc=3)
plt.grid(True)

plt.show()
```
